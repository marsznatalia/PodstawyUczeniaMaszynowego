2. Sprawdź, który z podziałów jest lepszy w przypadku miary entropii:
a) petal width (cm)na poziomie: 1.5
b) petal width (cm)na poziomie: 0.38
Podaj zysk informacyjny na każdym z podziałów

```python
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target

iris_counts = df['target'].value_counts()
print("Number of each iris type:\n", iris_counts)

total_irises = len(df)
print("\nTotal number of irises:", total_irises)
```

```python
import numpy as np
from collections import Counter

def entropy(labels):
    counts = Counter(labels)
    probabilities = [count / len(labels) for count in counts.values()]
    return -sum(p * np.log2(p) for p in probabilities if p > 0)
```

```python
group_a1 = df[df['petal width (cm)'] <= 1.5]['target']
group_a2 = df[df['petal width (cm)'] > 1.5]['target']
```

```python
print(len(group_a1))
print(len(group_a2))
```

```python
entropy_parent = entropy(df['target'])
print("Entropy of Parent:", entropy_parent)
```

```python
entropy_children_a1 = entropy(group_a1)
entropy_children_a2 = entropy(group_a2)
print("Entropy of Children A1:", entropy_children_a1)
print("Entropy of Children A2:", entropy_children_a2)
```

```python
average_entropy_a = (len(group_a1) / len(df) * entropy_children_a1) + (len(group_a2) / len(df) * entropy_children_a2)
#liczba danej kategorii do liczby wszystkich elementów w tej grupie i to jest suma
```

```python
information_gain_a = entropy_parent - average_entropy_a
print("Information Gain for Group A:", information_gain_a)
```

```python
group_b1 = df[df['petal width (cm)'] <= 0.38]['target']
group_b2 = df[df['petal width (cm)'] > 0.38]['target']
```

```python
entropy_children_b1 = entropy(group_b1)
entropy_children_b2 = entropy(group_b2)
```

```python
average_entropy_b = (len(group_b1) / len(df) * entropy_children_b1) + (len(group_b2) / len(df) * entropy_children_b2)
```

```python
information_gain_b = entropy_parent - average_entropy_b
print("Information Gain for Group B:", information_gain_b)
```

Większy zysk informacyjny sugeruje niższą entropię grupy lub grup próbek, a tym
samym mniejsze zaskoczenie.

```python
import sklearn.ensemble
import sklearn.tree
sklearn.ensemble.RandomForestClassifier
sklearn.tree.DecisionTreeClassifier
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

X = df.drop('target', axis=1)
y = df['target']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
```

```python
decision_tree = sklearn.tree.DecisionTreeClassifier(criterion='entropy')
decision_tree.fit(X_train, y_train)
y_pred_dt = decision_tree.predict(X_test)
accuracy_dt = accuracy_score(y_test, y_pred_dt)
print(f"Dokładność drzewa decyzyjnego: {accuracy_dt}")
```

```python
random_forest = sklearn.ensemble.RandomForestClassifier(n_estimators=15)
random_forest.fit(X_train, y_train)
y_pred_rf = random_forest.predict(X_test)
accuracy_rf = accuracy_score(y_test, y_pred_rf)
print(f"Dokładność lasu losowego: {accuracy_rf}")
```
